# TDCOSMO measurement comparison
This notebook generates a plot to compare the different measurements of the TDCOSMO collaboration. The notebook is adopted from the [H0 tension notebook](https://github.com/shsuyu/H0LiCOW-public/blob/master/H0_tension_plots/H0_tension.ipynb).

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib import rcParams

rcParams["text.usetex"] = False
rcParams["font.family"] = "sans-serif"
rcParams["font.sans-serif"] = ["Helvetica"]
rcParams["mathtext.fontset"] = "dejavusans"


measurement_list = []
```

## PL or  NFW + stars/constant M/L
Previous measurements of the TDCOSMO/H0LiCOW collaboration by
[Wong et al. 2020](https://ui.adsabs.harvard.edu/abs/2020MNRAS.tmp.1661W/abstract),
[Millon et al. 2020](https://ui.adsabs.harvard.edu/abs/2019arXiv191208027M/abstract)
and 
[Birrer et al. 2020](https://ui.adsabs.harvard.edu/abs/2020A%26A...643A.165B/abstract).

```python
label_h0licow = 'H0LiCOW (average of PL and NFW + stars/constant M/L)'
h0licow_median, h0licow_low, h0licow_up = 73.3, 1.8, 1.7
measurement_list.append([label_h0licow, h0licow_median, h0licow_low, h0licow_up])


label_nfw_stars = 'TDCOSMO-1 (NFW + stars/constant M/L)'
nfw_stars_median, nfw_stars_low, nfw_stars_up = 74.0, 1.8, 1.7
measurement_list.append([label_nfw_stars, nfw_stars_median, nfw_stars_low, nfw_stars_up])


label_pl = 'TDCOSMO-1 (power-law)'
pl_median, pl_low, pl_up = 74.2, 1.6, 1.6
measurement_list.append([label_pl, pl_median, pl_low, pl_up])

label_mst_tdcosmo_2020 = 'TDCOSMO-4'
mst_tdcosmo_median, mst_tdcosmo_low, mst_tdcosmo_up = 74.5, 6.1, 5.6
measurement_list.append([label_mst_tdcosmo_2020, mst_tdcosmo_median, mst_tdcosmo_low, mst_tdcosmo_up])

label_mst_tdcosmo_2020_slacs_ifu = r'TDCOSMO-4 + SLACS' #; self-similarity between samples)'
mst_tdcosmo_slacs_ifu_median, mst_tdcosmo_slacs_ifu_low, mst_tdcosmo_slacs_ifu_up = 67.4, 3.2, 4.1
measurement_list.append([label_mst_tdcosmo_2020_slacs_ifu, mst_tdcosmo_slacs_ifu_median, mst_tdcosmo_slacs_ifu_low, mst_tdcosmo_slacs_ifu_up])

```

## Hierarchical analysis with MST (this work)

We use the combination with Pantheon+

```python
Sn_sample = 'Pantheon+'

label_mst_tdcosmo = 'TDCOSMO-2025 + ' + Sn_sample
if Sn_sample == 'DES-SN5YR':
    mst_tdcosmo_median, mst_tdcosmo_low, mst_tdcosmo_up = 71.2, 3.6, 3.7
elif Sn_sample == 'Pantheon+':
    mst_tdcosmo_median, mst_tdcosmo_low, mst_tdcosmo_up = 71.6, 3.3, 3.9
measurement_list.append([label_mst_tdcosmo, mst_tdcosmo_median, mst_tdcosmo_low, mst_tdcosmo_up])

label_mst_tdcosmo_slacs= r'TDCOSMO-2025 + SLACS + ' + Sn_sample
if Sn_sample == 'DES-SN5YR':
    mst_tdcosmo_ifu_median, mst_tdcosmo_ifu_low, mst_tdcosmo_ifu_up = 73.8, 3.6,3.2
elif Sn_sample == 'Pantheon+':
    mst_tdcosmo_ifu_median, mst_tdcosmo_ifu_low, mst_tdcosmo_ifu_up = 74.1,3.3,3.6
measurement_list.append([label_mst_tdcosmo_slacs, mst_tdcosmo_ifu_median, mst_tdcosmo_ifu_low, mst_tdcosmo_ifu_up])

label_mst_tdcosmo_sl2s = r'TDCOSMO-2025 + SL2S + ' + Sn_sample # ; self-similarity between samples)'
if Sn_sample == 'DES-SN5YR':
    mst_tdcosmo_sl2s_median, mst_tdcosmo_sl2s_low, mst_tdcosmo_sl2s_up = 71.7,3.5,3.6
elif Sn_sample == 'Pantheon+':
    mst_tdcosmo_sl2s_median, mst_tdcosmo_sl2s_low, mst_tdcosmo_sl2s_up = 72.1,3.4,3.6
measurement_list.append([label_mst_tdcosmo_sl2s, mst_tdcosmo_sl2s_median, mst_tdcosmo_sl2s_low, mst_tdcosmo_sl2s_up])

label_mst_tdcosmo_slacs_sl2s = r'TDCOSMO-2025 + SLACS + SL2S + ' + Sn_sample # ; self-similarity between samples)'
if Sn_sample == 'DES-SN5YR':
    mst_tdcosmo_slacs_sl2s_median, mst_tdcosmo_slacs_sl2s_low, mst_tdcosmo_slacs_sl2s_up = 73.9,3.0,3.4
elif Sn_sample == 'Pantheon+':
    mst_tdcosmo_slacs_sl2s_median, mst_tdcosmo_slacs_sl2s_low, mst_tdcosmo_slacs_sl2s_up = 74.3,3.7,3.1
measurement_list.append([label_mst_tdcosmo_slacs_sl2s, mst_tdcosmo_slacs_sl2s_median, mst_tdcosmo_slacs_sl2s_low, mst_tdcosmo_slacs_sl2s_up])
```

```python

# Special fine-tuned design, adapted from Bonvin & Millon notebook

ys = [14, 11, 9, 4, 1.5, -4.0, -6.5, -9, -11.5]

plt.figure(figsize=(8.0, 10), dpi=200)
plt.subplots_adjust(left=0.01, right=0.99, bottom=0.20, top=0.92)

ax = plt.subplot(1, 1, 1)
plt.xlabel(r"$H_0\rm{\ [km\,s^{-1}\,Mpc^{-1}]}$", fontsize=24)
plt.yticks(fontsize=14)
plt.xticks(fontsize=20)


colors = ['peru', 'goldenrod', 'goldenrod','royalblue', 'mediumvioletred', 'coral', 'rebeccapurple', 'crimson','purple',]


plt.axis([59.1, 82.9, -13.5, 15.5]) # 75.6
plt.gca().yaxis.set_major_locator(plt.NullLocator())
plt.title(r'$H_0$ measurements in $\rm{flat\,\, \Lambda CDM}$ - performed blindly', fontsize=17)

ind = np.arange(len(measurement_list))
for x, y, c, i in zip(measurement_list, ys, colors, ind):
    pcs = [x[1]-x[2], x[1], x[1]+x[3]]
    label = x[0]
    if i == 4 :
        linestyle = '--'
    else :
        linestyle = '-'

    plt.scatter(pcs[1], y, s=80, edgecolor=c, c=c, label=label)
    plt.plot([pcs[0], pcs[2]], [y, y], linewidth=2.5, color=c, linestyle=linestyle)

    delaytext = r"$%.1f^{+%.1f}_{-%.1f}$" % (pcs[1], pcs[2]-pcs[1], pcs[1]-pcs[0])
    plt.annotate(delaytext, xy=(pcs[1], y + 0.3), color = c, horizontalalignment="center", fontsize=17)
    plt.annotate(label, xy=(pcs[1], y-0.7), color = "dimgrey", horizontalalignment="center", fontsize=10) 

    
plt.axhline(12.5, color="darkgrey", ls='--', alpha=0.5)  
plt.axhline(7.5, color="k", ls='--', alpha=0.5)  
plt.axhline(0, color="darkgrey", ls='--', alpha=0.5)  

bbox_props = dict(boxstyle="square", lw=1,  fc="white")
plt.annotate("Wong et al. 2020", xy=(59.5, 14.5), color = "black", 
             horizontalalignment="left", verticalalignment="center", fontsize=13, bbox=bbox_props)
plt.annotate("6 time-delay lenses", xy=(59.5, 13.5), color = "dimgrey", 
             horizontalalignment="left", verticalalignment="center", fontsize=10)

plt.annotate("Millon et al. 2020", xy=(59.5, 11.5), color = "black", 
             horizontalalignment="left", verticalalignment="center", fontsize=13, bbox=bbox_props)
plt.annotate("7 time-delay lenses (6 H0LiCOW + 1 STRIDES)", xy=(59.5, 10.5), color = "dimgrey", 
             horizontalalignment="left", verticalalignment="center", fontsize=10)
plt.annotate("Birrer et al. 2020", xy=(59.5, 6.5), color = "black", 
             horizontalalignment="left", verticalalignment="center", fontsize=13, bbox=bbox_props,)
plt.annotate("7 time-delay lenses (+ 33 SLACS lenses)", xy=(59.5, 5.5), color = "dimgrey", 
             horizontalalignment="left", verticalalignment="center", fontsize=10)
plt.annotate("kinematics-only constraints on mass profile", xy=(72, 6.5), color = "black", 
             horizontalalignment="center", verticalalignment="center", fontsize=13)

plt.annotate("This work", xy=(59.5, -1.0), color = "black", 
             horizontalalignment="left", verticalalignment="center", fontsize=13, bbox=bbox_props)
plt.annotate("8 time-delay lenses (+ 11 SLACS lenses + 4 SL2S) with improved kinematics", xy=(59.5, -2.0), color = "dimgrey", 
             horizontalalignment="left", verticalalignment="center", fontsize=10)

    
plt.tight_layout()
plt.savefig(f'./figures/h0_comparison_milestone2025_{Sn_sample}.pdf')
plt.show()
```
